# Tables and pictures

A query can show more than its bindings. `jupyter:print_table/1` shows every solution of a goal, one to a row.

```prolog
jupyter:print_table((between(1, 6, N), Square is N * N, Power is 2 ^ N)).
```

A cell that starts with `%%table` shows each of its queries as a table, and one that starts with `%%csv FILE` writes the table to a file as well, which the file browser shows.

```prolog
%%csv powers.csv
between(1, 6, N), Power is 2 ^ N.
```

`jupyter:display/2` shows a text as its MIME type says: Markdown, HTML or SVG among them.

```prolog
jupyter:display('text/markdown', 'Some **bold** text, and a list:\n\n- one\n- two').
```

A program can draw. This one draws a bar for each pair of a list.

```prolog
%%user
chart(Pairs) :-
    length(Pairs, N),
    Height is N * 24,
    with_output_to(atom(Svg), (
        format('<svg xmlns="http://www.w3.org/2000/svg" width="420" height="~w">', [Height]),
        forall(nth0(I, Pairs, Label-Value),
               (   Y is I * 24, Baseline is Y + 14, Width is Value * 4,
                   format('<text x="0" y="~w" font-size="13">~w</text>', [Baseline, Label]),
                   format('<rect x="40" y="~w" width="~w" height="18" fill="#34d399"/>', [Y, Width])
               )),
        format('</svg>', []))),
    jupyter:display('image/svg+xml', Svg).
```

```prolog
findall(N-Square, (between(1, 8, N), Square is N * N), Pairs), chart(Pairs).
```
