# Grammars

A definite clause grammar describes a language. `phrase/2` parses a list with it, and the same grammar gives the sentences of the language.

```prolog
%%user
greeting --> [hello], name.
name --> [world].
name --> [prolog].
```

```prolog
phrase(greeting, [hello, Who]).
```

**Next**, under an answer that ends with ` ;`, shows the next solution of its query, and so does a cell that is `retry.`

```prolog
retry.
```

```prolog
findall(Sentence, phrase(greeting, Sentence), Sentences).
```

## Arithmetic

A text in double quotes is a list of characters, so a grammar reads it a character at a time. This one computes the value of what it reads, with `*` binding tighter than `+` and `-`.

```prolog
%%user+
expr(V) --> term(T), expr_rest(T, V).
expr_rest(A, V) --> "+", !, term(T), { B is A + T }, expr_rest(B, V).
expr_rest(A, V) --> "-", !, term(T), { B is A - T }, expr_rest(B, V).
expr_rest(V, V) --> [].
term(V) --> factor(F), term_rest(F, V).
term_rest(A, V) --> "*", !, factor(F), { B is A * F }, term_rest(B, V).
term_rest(V, V) --> [].
factor(V) --> "(", !, expr(V), ")".
factor(V) --> digits(Ds), { number_chars(V, Ds) }.
digits([D|Ds]) --> digit(D), digits_rest(Ds).
digits_rest([D|Ds]) --> digit(D), !, digits_rest(Ds).
digits_rest([]) --> [].
digit(D) --> [D], { char_code(D, C), C >= 0'0, C =< 0'9 }.
```

```prolog
phrase(expr(V), "2*(3+4)-5").
```

```prolog
findall([Text, V], (member(Text, ['1+2*3', '(1+2)*3', '10-2-3']), atom_chars(Text, Cs), phrase(expr(V), Cs)), _Rows),
jupyter:print_table(_Rows, ['Text', 'Value']).
```

A text that is not an expression has no parse.

```prolog
\+ phrase(expr(_), "2+").
```
