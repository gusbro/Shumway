# Graphs

A query can draw a graph: a term as the tree of its subterms, or the edges a predicate gives. The server writes the graph in DOT, and the notebook draws it.

## Terms

`show_term/1` draws a term: a node for each subterm, an edge from a term to each of its arguments, numbered by position. Atoms, numbers and compound terms have shapes of their own.

```prolog
show_term(person(ann, born(1990, may), [prolog, logic])).
```

`%%tree` draws the term of the cell, here an arithmetic expression as the reader sees it.

```prolog
%%tree
1 + 2 * 3 - 4
```

## Transitions

A predicate whose solutions are edges draws as a graph. `jupyter:print_transition_graph(Pred/Arity, From, To, Label)` takes the edge from the argument `From` to the argument `To`, labelled by the argument `Label`, or by nothing when it is 0.

```prolog
%%user
step(idle, coin, ready).
step(ready, push, idle).
step(ready, coin, ready).
step(idle, push, idle).
```

```prolog
jupyter:print_transition_graph(step/3, 1, 3, 2).
```

## Styled nodes and edges

`show_graph/2` takes the nodes too: the first argument of the node predicate names a node, the second lists its DOT attributes. A label argument that is a list gives an edge its label, colour and style.

```prolog
%%user
state(idle, [label/'Idle', shape/box]).
state(ready, [label/'Ready', shape/box, style/filled, fillcolor/lightyellow]).
move(idle, [label/coin, color/blue], ready).
move(ready, [label/push, color/red, style/dashed], idle).
```

```prolog
show_graph(state/2, move/3).
```
