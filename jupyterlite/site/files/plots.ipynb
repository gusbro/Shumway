# Plots

A query draws a plot with `show_data/1`, or a `%%data` cell: a goal that binds `Data` to a list of `Key-Value` pairs. `type` names the kind of plot; the other keys are those of the matplotlib function of that name. An option the notebook does not draw is named below the plot.

## Bars

```prolog
%%data
Data = [type-bar, title-'Fruit sold', x-[apples, pears, plums, figs], height-[12, 7, 9, 4],
        color-[orangered, gold, purple, olive], bar_label-[label_type-edge]]
```

```prolog
%%data
Data = [type-barh, title-'Hours a week', y-[reading, coding, walking], width-[5, 20, 7],
        color-steelblue]
```

```prolog
%%data
findall(X, (between(1, 200, I), X is sin(I) * 10 + cos(I * 7) * 3), Xs),
Data = [type-hist, title-'Histogram', x-Xs, bins-15, color-skyblue, edgecolor-black]
```

## Lines and points

```prolog
%%data
findall(X, between(0, 8, X), Xs),
findall(Y, (member(X, Xs), Y is X * X), Ys),
Data = [type-plot, title-'Squares', x-Xs, y-Ys, marker-o, color-darkgreen,
        xlabel-'n', ylabel-'n squared', grid-true]
```

```prolog
%%data
Data = [type-scatter, title-'Scatter', x-[5, 7, 8, 7, 2, 17, 2, 9, 4, 11],
        y-[99, 86, 87, 88, 111, 86, 103, 87, 94, 78], color-crimson]
```

```prolog
%%data
Data = [type-step, title-'Steps', x-[1, 2, 3, 4, 5], y-[0, 1, 0, 2, 1], where-post]
```

```prolog
%%data
findall(X, between(0, 18, X), Xs),
findall(Y, (member(X, Xs), Y is sin(X / 3)), Ys),
Data = [type-stem, title-'Stem', x-Xs, y-Ys]
```

```prolog
%%data
Data = [type-loglog, title-'Powers of two', x-[1, 2, 4, 8, 16, 32], y-[2, 4, 16, 256, 65536, 4294967296]]
```

## Shares, stacks and spreads

```prolog
%%data
Data = [type-pie, title-'Languages', x-[45, 30, 15, 10], labels-['Prolog', 'Logtalk', 'C#', 'Other'],
        autopct-'%1.1f%%']
```

```prolog
%%data
Data = [type-stackplot, title-'Hours', x-[1, 2, 3, 4, 5], y-[[7, 8, 6, 11, 7], [8, 5, 7, 8, 13]],
        labels-['Studying', 'Playing'], legend-[loc-'upper left'], colors-[orange, teal]]
```

```prolog
%%data
Data = [type-errorbar, title-'Measurements', x-[1, 2, 3, 4], y-[2.1, 3.9, 6.2, 7.8], yerr-[0.3, 0.5, 0.4, 0.6]]
```

```prolog
%%data
Data = [type-boxplot, title-'Spread', x-[[1, 2, 2, 3, 4, 9], [3, 4, 4, 5, 5, 6]], labels-[first, second]]
```

```prolog
%%data
Data = [type-ecdf, title-'Cumulative', x-[3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]]
```

```prolog
%%data
Data = [type-eventplot, title-'Events', positions-[[1, 3, 4, 7], [2, 5, 6, 8.5]]]
```
