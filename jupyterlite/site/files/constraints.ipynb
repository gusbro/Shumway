# Constraints

`library(clpfd)` solves problems over the integers. A constraint narrows the values a variable can take, and what is left of them is the answer.

```prolog
use_module(library(clpfd)).
```

```prolog
X #> 3, X #< 6.
```

`label/1` gives the values one at a time, and `retry.` the next one.

```prolog
X in 1..10, X #> 7, label([X]).
```

```prolog
retry.
```

## The length of a list

`@L` is the length of the list `L`, and a constraint says something of it wherever it says something of an integer. A fixed length closes the list.

```prolog
@L #= 3.
```

A bound on the length makes the list at least that long, and `label/1` gives the lengths one at a time.

```prolog
@L #< 3, label([@L]).
```

```prolog
retry.
```

Two lists can be related through their lengths.

```prolog
@A #= 2 * @B, B = [ann, bob].
```

A length is a variable like any other. These are the lists of two or three bits with two ones.

```prolog
jupyter:print_table((@Bits #= N, N in 2..3, label([N]), Bits ins 0..1, sum(Bits, #=, 2), label(Bits))).
```

## A puzzle

SEND + MORE = MONEY, where each letter is a different digit and no number starts with 0.

```prolog
%%user
:- use_module(library(clpfd)).

puzzle([S,E,N,D] + [M,O,R,E] = [M,O,N,E,Y]) :-
    Vars = [S,E,N,D,M,O,R,Y],
    Vars ins 0..9,
    all_different(Vars),
    S*1000 + E*100 + N*10 + D + M*1000 + O*100 + R*10 + E #=
        M*10000 + O*1000 + N*100 + E*10 + Y,
    M #\= 0, S #\= 0,
    label(Vars).
```

## Six queens

No two queens share a row, a column or a diagonal. `labeling([ff], Qs)` tries first the queen with the fewest places left. The table has every solution: the row of the queen in each column.

```prolog
%%user+
queens(N, Qs) :-
    length(Qs, N),
    Qs ins 1..N,
    safe(Qs),
    labeling([ff], Qs).

safe([]).
safe([Q|Qs]) :- no_attack(Q, Qs, 1), safe(Qs).

no_attack(_, [], _).
no_attack(Q, [Q1|Qs], D) :-
    Q #\= Q1,
    abs(Q - Q1) #\= D,
    D1 is D + 1,
    no_attack(Q, Qs, D1).
```

```prolog
jupyter:print_table(queens(6, Qs)).
```
